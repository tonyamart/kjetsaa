```python
df = pd.read_csv('../data/poetree_ru_19c_tokens.csv')
df.head()
```

```python
len(df)
```

```python
# check if some of the cells are in wrong type / empty
df = df.loc[df.word.apply(lambda x: not isinstance(x, (float, int)))]
```

```python
len(df)
```

```python
# lemmatization
df['lemma'] = df['word'].apply(m.lemmatize)
df['lemma'] = [''.join(map(str, l)) for l in df['lemma']] 
df['lemma'] = df['lemma'].str.replace('\n', '', regex = True)
```

```python
# run analysis
df['analysis'] = df['word'].apply(m.analyze)
```

```python
df.tail()
```

```python
df = df.drop('Unnamed: 0', axis = 1)
# df = df.drop('text', axis = 1)
```
